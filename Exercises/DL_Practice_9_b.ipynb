```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.a = nn.Linear(10, 10)
        self.b = nn.Linear(10, 10)
    
    def forward(self, x):
        return self.a(x) + self.b(x)

model: MyModel = MyModel()
len(list(model.parameters()))
```

```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.submodules = {
            "a": nn.Linear(10, 10),
            "b": nn.Linear(10, 10),
        }
    
    def forward(self, x):
        return self.submodules["a"](x) + self.submodules["b"](x)

model: MyModel = MyModel()
len(list(model.parameters()))
```

```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.submodules = nn.ModuleDict(
            {
                "a": nn.Linear(10, 10),
                "b": nn.Linear(10, 10),
            }
        )
    
    def forward(self, x):
        return self.submodules["a"](x) + self.submodules["b"](x)

model: MyModel = MyModel()
len(list(model.parameters()))
```
